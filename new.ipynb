```python
import json
from bs4 import BeautifulSoup
```

```python
# Load the JSON data
with open('crew_scraped_data.json', 'r') as f:
    data = json.load(f)
```

```python
# Function to clean HTML content
def clean_html(html_content):
    """Keep only <h1>, <h2>, and <p> tags in the HTML content."""
    soup = BeautifulSoup(html_content, 'html.parser')
    
    # Find all <h1>, <h2>, and <p> tags
    allowed_tags = soup.find_all(['h1', 'h2', 'p'])
    
    # Create a new BeautifulSoup object to store the cleaned content
    cleaned_soup = BeautifulSoup('', 'html.parser')
    
    # Add the allowed tags to the cleaned soup
    for tag in allowed_tags:
        cleaned_soup.append(tag)
    
    return str(cleaned_soup)
```

```python
# Clean the HTML content in the JSON data
for article in data['data']:
    article['content'] = clean_html(article['content'])
```

```python
# Save the cleaned data to a new JSON file
with open('cleaned_crew_scraped_data.json', 'w') as f:
    json.dump(data, f, indent=4)

print("HTML content cleaned and saved to 'cleaned_crew_scraped_data.json'.")
```
